```python
# convert csv file to json file, where columns are keys, and rows are lists in the values
import pandas as pd
```

```python
csv_path = 'data.csv'
```

```python
# read csv file
df = pd.read_csv(csv_path)

# randomly shuffle the rows with seed 42
df = df.sample(frac=1, random_state=42)

df
```

```python
# turn the column name creativity_score to label
df.rename(columns={'creativity_score': 'label'}, inplace=True)
```

```python
# split by 3:1:1 for train, validation, test
train_df = df.iloc[:int(len(df)*0.6)]
val_df = df.iloc[int(len(df)*0.6):int(len(df)*0.8)]
test_df = df.iloc[int(len(df)*0.8):]
```

```python
test_df
```

```python
train_json_data = {}
for col in df.columns:
    train_json_data[col] = train_df[col].tolist()

val_json_data = {}
for col in df.columns:
    val_json_data[col] = val_df[col].tolist()

test_json_data = {}
for col in df.columns:
    test_json_data[col] = test_df[col].tolist()
```

```python
train_json_path = 'adventure_train.json'
val_json_path = 'adventure_val.json'
test_json_path = 'adventure_test.json'

import json

with open(train_json_path, 'w') as f:
    json.dump(train_json_data, f)

with open(val_json_path, 'w') as f:
    json.dump(val_json_data, f)

with open(test_json_path, 'w') as f:
    json.dump(test_json_data, f)
```

```python
import numpy as np
from scipy import stats

# Sample data
data = train_json_data['label']

# max and min
max_val = max(data)
min_val = min(data)
print(f"Max: {max_val}, Min: {min_val}")

# Calculate the mean (mu)
mu = np.mean(data)
print(f"Mean (mu): {mu}")

# Calculate the standard deviation (sigma)
sigma = np.std(data)

# Calculate the z-score using the formula: (x - mu) / sigma
z_scores = (data - mu) / sigma
print(f"Z-scores (manual calculation): {z_scores}")

# Calculate the z-score using SciPy's zscore function
z_scores_scipy = stats.zscore(data)
print(f"Z-scores (using scipy.stats.zscore): {z_scores_scipy}")
```
